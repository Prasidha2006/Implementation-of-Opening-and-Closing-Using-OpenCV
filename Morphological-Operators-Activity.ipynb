```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
def load_img():
    blank_img =np.zeros((600,600))
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(blank_img,text='ABCDE',org=(50,300), fontFace=font,fontScale= 5,color=(255,255,255),thickness=25,lineType=cv2.LINE_AA)
    return blank_img
```

```python
def display_img(img):
    fig = plt.figure(figsize=(12,10))
    ax = fig.add_subplot(111)
    ax.imshow(img,cmap='gray')
    plt.show()
```

```python
img = load_img()
```

```python
kernel = np.ones((15,15), np.uint8)
```

```python
# Opening = Erosion followed by Dilation
opening = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
```

```python
noise = img.copy()

for i in range(100):
    x = np.random.randint(0, 600)
    y = np.random.randint(0, 600)
    cv2.circle(noise, (x, y), 3, 255, -1)
```

```python
plt.figure(figsize=(15,5))
plt.subplot(131)
plt.imshow(img, cmap='gray')
plt.title("Original")

plt.subplot(132)
plt.imshow(noise, cmap='gray')
plt.title("Noisy Image")

plt.subplot(133)
plt.imshow(opening, cmap='gray')
plt.title("Opening")
```

```python
hole_img = img.copy()

cv2.circle(hole_img, (180,250), 5, 0, -1)
cv2.circle(hole_img, (300,250), 5, 0, -1)
cv2.circle(hole_img, (420,250), 5, 0, -1)
```

```python
kernel = np.ones((7,7), np.uint8)

closing = cv2.morphologyEx(
    hole_img,
    cv2.MORPH_CLOSE,
    kernel
)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(131)
plt.imshow(img, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(132)
plt.imshow(hole_img, cmap='gray')
plt.title("Image with Holes")
plt.axis('off')

plt.subplot(133)
plt.imshow(closing, cmap='gray')
plt.title("Closing")
plt.axis('off')

plt.show()
```
